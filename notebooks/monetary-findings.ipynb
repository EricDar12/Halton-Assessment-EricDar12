```python
import polars as pl
from utils import getConnection
con, dataset_path = getConnection()
```

```python
con.execute("""
    CREATE TEMP TABLE contracts AS
    SELECT *
    FROM read_csv_auto(
        '{0}',
        delim = ',',
        header = true,
        strict_mode = false,
        all_varchar = true
    )
""".format(dataset_path))
```

```python
con.sql("""
SELECT owner_org_title, COUNT(owner_org_title)
FROM contracts
WHERE owner_org_title LIKE 'Accessibility%'
GROUP BY owner_org_title
""")
```
